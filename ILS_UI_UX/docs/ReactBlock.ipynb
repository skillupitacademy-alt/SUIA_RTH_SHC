### The key React-specific rule

> **React tutorials should be more visual, interactive, state-flow-oriented, comparison-heavy, and architecture-oriented than ordinary JavaScript tutorials.**

And for your particular platform, I would make the Presentation Engine recognize these **React topic signals**:

```text
State
Hooks
Rendering
Reconciliation
Context
Events
Performance
Architecture
Routing
State Management
```

and automatically increase:

```text
DiagramBlock
+
QuestionBlock
+
InteractiveBlock
+
ComparisonBlock
+
InterviewBlock
+
BestPracticeBlock
```

```text
P7
P8
QZ8
IV7
INT6
```
